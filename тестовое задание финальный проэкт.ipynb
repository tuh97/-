Задание 2. DAU

DAU — количество уникальных пользователей, активных за один день.

Для расчёта данные были сгруппированы по дате с помощью groupby(). Затем для каждого дня подсчитано количество уникальных пользователей методом nunique().

Среднее значение DAU за ноябрь рассчитано методом mean().

Ответ: 560 пользователей.

```python
# Задание 3. Retention первого дня

# Пользователи, посетившие приложение 1 ноября
day_1 = set(
    audience.loc[
        audience["date"].dt.strftime("%Y-%m-%d") == "2023-11-01",
        "user_id"
    ]
)

# Пользователи, посетившие приложение 2 ноября
day_2 = set(
    audience.loc[
        audience["date"].dt.strftime("%Y-%m-%d") == "2023-11-02",
        "user_id"
    ]
)

# Пользователи, которые вернулись
returned = day_1.intersection(day_2)

# Расчёт Retention
retention = len(returned) / len(day_1) * 100

print("Пользователей 1 ноября:", len(day_1))
print("Вернулись 2 ноября:", len(returned))
print(f"Retention первого дня: {retention:.1f}%")
```

### Задание 3. Retention первого дня

Retention показывает долю пользователей, вернувшихся в приложение через определённое время.

Для расчёта были выбраны уникальные пользователи, посетившие приложение 1 ноября. Затем определено, сколько из них повторно посетили приложение 2 ноября.

**Формула:**

Retention = (Количество вернувшихся пользователей / Количество пользователей 1 ноября) × 100%

**Ответ: 26,6%.**

Это означает, что примерно 27 из 100 пользователей, посетивших приложение 1 ноября, вернулись на следующий день.

```python
# Задание 7. Расчёт NPS

total = 2000
promoters = 1200
detractors = 500

# Процент сторонников
promoters_percent = promoters / total * 100

# Процент критиков
detractors_percent = detractors / total * 100

# Рассчитываем NPS
nps = promoters_percent - detractors_percent

print("Сторонники:", promoters_percent, "%")
print("Критики:", detractors_percent, "%")
print("NPS:", nps, "%")
```

### Задание 7. Расчёт NPS

Сначала я посчитал процент сторонников и критиков от общего количества опрошенных.

Сторонников получилось 60%, а критиков — 25%.

Затем вычел процент критиков из процента сторонников:

NPS = 60% − 25% = 35%

**Ответ: 35%.**

Получается, что сторонников продукта больше, чем критиков. Это хороший результат.

### Задание 17. Что такое квартили?

Квартили — это значения, которые делят упорядоченные данные на четыре равные части.

Первый квартиль (Q1) соответствует 25% данных, второй (Q2) — 50%, а третий (Q3) — 75%.

Второй квартиль также называется медианой.

**Ответ: квартили делят данные на четыре равные части.**

```python
# Задание 18. Проверка A/B-теста

from statsmodels.stats.proportion import proportions_ztest

# Контрольная группа
control_users = 100047501
control_payments = 1003

# Тестовая группа
test_users = 100001055
test_payments = 1099

# Считаем конверсию
control_cr = control_payments / control_users * 100
test_cr = test_payments / test_users * 100

print(f"Конверсия контрольной группы: {control_cr:.5f}%")
print(f"Конверсия тестовой группы: {test_cr:.5f}%")

# Проверяем разницу с помощью Z-теста
payments = [control_payments, test_payments]
users = [control_users, test_users]

stat, p_value = proportions_ztest(payments, users)

print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("Разница статистически значима")
else:
    print("Статистически значимой разницы нет")
```

### Задание 18. Проверка A/B-теста

Сначала я посчитал конверсию в оплату для двух групп.

В контрольной группе получилось примерно 0,00100%, а в тестовой — 0,00110%.

Потом я использовал Z-тест, чтобы проверить разницу между конверсиями.

Получилось p-value около 0,036. Это меньше 0,05, поэтому разница статистически значима.

**Вывод:** в тестовой группе конверсия выше примерно на 9,6%. Изменение показало положительный результат. Перед внедрением стоит также проверить расходы и другие показатели.

```python
# Задание 8. Итоговые результаты A/B-тестов

from scipy.stats import ttest_ind

for experiment in sorted(ab_tests["experiment_num"].unique()):

    data = ab_tests[ab_tests["experiment_num"] == experiment]

    groups = sorted(data["experiment_group"].dropna().unique())

    group_a = data.loc[
        data["experiment_group"] == groups[0], "revenue"
    ].dropna()

    group_b = data.loc[
        data["experiment_group"] == groups[1], "revenue"
    ].dropna()

    p_value = ttest_ind(
        group_a, group_b, equal_var=False
    ).pvalue

    print(f"Эксперимент № {experiment}")
    print(f"Средний доход группы {groups[0]}: {group_a.mean():.2f}")
    print(f"Средний доход группы {groups[1]}: {group_b.mean():.2f}")
    print(f"p-value: {p_value:.4f}")

    if p_value < 0.05:
        print("Разница статистически значима")
    else:
        print("Значимой разницы нет")

    print("-" * 40)
```
